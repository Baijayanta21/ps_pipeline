```python
import myutils.tge.grid as gd
```

```python
# Set the gridding parameters in the TGE code
n1       = 0     # Starting channel number
n2       = 767   # End channel number  
Umax     = 250   # Baselines greater than this are rejected
FWHM     = 23    # Primary beam FWHM of the telescope in degrees
f        = 0.6   # Tapering parameter value
Flag     = True  # Apply actual flagging of data
nstokes  = [0,1] # 0 for XX and 1 for YY (which polarizations you want to grid)
```

```python
infile   = '/home/MWA_data/1000007200.fits' # fits file name
```

### $\textrm{Apply to Data}$

```python
nrel     = -1  # data file
GV, info = gd.grid(infile, n1, n2, nrel, Umax, FWHM, f, Flag, nstokes)
print(f"{GV.shape = }")
```

```python
print(f"dU   : {info[0]:.2f}\
      \nUmax : {info[1]}\
      \nFWHM : {info[2]} degrees\
      \nf    : {info[3]}")
```

```python
import os
index = int(os.path.split(infile)[1].split(".")[0][-4:])
print(index)
```

```python
import numpy as np
np.save(f'GV_{index}.npy',GV) # save the GV file as .npy array
```

### $\textrm{Apply to Noise}$

```python
nrel     = -2  # noise
seed_val = 7200
GVn      = gd.grid(infile, n1, n2, nrel, Umax, FWHM, f, Flag, nstokes, seed = seed_val)[0]
print(f"{GVn.shape = }")
```

```python
np.save(f'GV_{index}_Noise_{seed_val}.npy',GVn) # save the GV file as .npy array
# file name format f'GV_{index}_Noise_{seed_val}.npy'
```
