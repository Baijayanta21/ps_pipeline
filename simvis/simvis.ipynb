```python
import simvis as sv
```

```python
import builtins as blt

# Set simulation parameters
blt.nside = 512        # nside parameter
blt.Nrea  = 100        # no of realizations
blt.chunk_size = 500   # chunk_size, no of baselines in a single step
```

```python
os.makedirs("./simulated/", exist_ok=True)
```

```python
infits        = '/home/MWA_data/1000007200.fits'     # data file name
out_uaps_fits = './simulated/1000007200_uaps.fits'          # output uaps fits file name
out_aps_fits  = './simulated/1000007200_aps.fits'           # output  aps fits file name
out_ps_fits   = './simulated/1000007200_ps.fits'            # output   ps fits file name
```

```python
# simulate APS 100 realizations
sv.sim_vis(infits, out_aps_fits,  skysimtype = '2D', seed = 15)
```

```python
# simulate UAPS 100 realizations
sv.sim_vis(infits, out_uaps_fits, skysimtype = '2D', seed = 15, apsfunc = sv.uaps)
```

```python
# set a new chunk_size for 3D simulation
blt.chunk_size = 700   # chunk_size, no of baselines in a single step
```

```python
# simulate 3D visibilities 
sv.sim_vis(infits, out_ps_fits,   skysimtype = '3D', seed = 15)
```
